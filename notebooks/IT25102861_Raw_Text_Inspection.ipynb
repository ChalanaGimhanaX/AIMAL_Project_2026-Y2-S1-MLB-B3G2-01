```python
df = pd.read_csv("../data/raw/bbc-text.csv")
```

```python
print(df.duplicated().sum())
```

```python
#99 duplicate rows found
df = df.drop_duplicates()
print(f"After removing: {df.shape[0]} rows")
```

```python
df['category'].value_counts()
```

```python
df['category'].value_counts().plot(kind='bar', color='steelblue')
plt.title('Category Distribution')
plt.ylabel('Count')
plt.tight_layout()
```

```python
#check article lengths
```

```python
df['word_count'] = df['text'].apply(lambda x: len(str(x).split()))
df['word_count'].describe()
```

```python
plt.hist(df['word_count'], bins=30, color='steelblue', edgecolor='white')
plt.title('Article Word Count Distribution')
plt.xlabel('Word Count')
plt.ylabel('Frequency')
```
