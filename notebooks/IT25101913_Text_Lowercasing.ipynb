```python
import pandas as pd
import re
```

```python
df = pd.read_csv("../results/outputs/step1_inspected_text.csv")
```

```python
df.head()
```

```python
#check if text has any uppercase letters
```

```python
uppercase_count = df['text'].apply(lambda x: len(re.findall(r'[A-Z]', str(x)))).sum()
print(f"Total uppercase characters: {uppercase_count}")
```

```python
#this kaggle dataset is alredy lowercased
#but we still apply .lower() for pipeline safety
```

```python
df['text'] = df['text'].apply(lambda x: str(x).lower())
```

```python
#verify no uppercase remaining
remaining = df['text'].apply(lambda x: len(re.findall(r'[A-Z]', str(x)))).sum()
print(f"Uppercase after lowercasing: {remaining}")
```

```python
#to show the effect lets use some raw text with uppercase
```

```python
sample_text = "The BBC reported that Apple Inc shares ROSE 5% on Monday."
print("Before:", sample_text)
print("After:", sample_text.lower())
```
