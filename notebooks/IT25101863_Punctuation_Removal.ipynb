```python
import pandas as pd
import re
```

```python
df = pd.read_csv("../results/outputs/step2_lowercased_text.csv")
```

```python
df.head()
```

```python
#check how much punctuation is in the text
```

```python
punct_counts = df['text'].apply(lambda x: len(re.findall(r'[^a-z0-9\s]', str(x))))
print(f"Total punctuation characters: {punct_counts.sum()}")
print(f"Average per article: {punct_counts.mean():.1f}")
```

```python
#show sample before cleaning
```

```python
print(df['text'].iloc[0][:200])
```

```python
#remove punctuation and special characters
def remove_punctuation(text):
    return re.sub(r'[^a-z0-9\s]', ' ', str(text))
```

```python
df['text'] = df['text'].apply(remove_punctuation)
```

```python
#check after removal
print(df['text'].iloc[0][:200])
```

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(punct_counts, bins=30, color='mediumpurple', edgecolor='white')
plt.title('Punctuation Count per Article')
plt.xlabel('Punctuation Count')
plt.ylabel('Frequency')
```
