```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../results/outputs/step4_stopwords_removed.csv")
```

```python
df.head()
```

```python
#tokenize the text
```

```python
df['tokens'] = df['text'].apply(lambda x: str(x).split())
```

```python
df[['category','tokens']].head()
```

```python
#check token counts per article
```

```python
df['token_count'] = df['tokens'].apply(len)
df['token_count'].describe()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df['token_count'], bins=30, color='steelblue')
plt.title('Token Count Distribution')
plt.show()
```

```python
#create clean text column
```

```python
df['clean_text'] = df['text']
```

```python
#save final preprocessed text
```

```python
df[['category','clean_text']].to_csv('../results/outputs/step5_tokenized_text.csv', index=False)
print("saved")
print(f"Total articles: {len(df)}")
```
