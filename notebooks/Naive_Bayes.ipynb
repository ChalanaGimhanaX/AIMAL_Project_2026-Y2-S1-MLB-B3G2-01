# Naive Bayes Model Training - BBC News Classification
Train and evaluate Multinomial Naive Bayes on TF-IDF features.

```python
import pandas as pd
import joblib
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train_df = pd.read_csv('../results/outputs/train_data.csv')
test_df = pd.read_csv('../results/outputs/test_data.csv')

print('Train shape:', train_df.shape)
print('Test shape:', test_df.shape)
```

```python
# Vectorize using TF-IDF
tfidf = TfidfVectorizer(max_features=5000)
X_train = tfidf.fit_transform(train_df['clean_text'])
X_test = tfidf.transform(test_df['clean_text'])

y_train = train_df['category']
y_test = test_df['category']

print('TF-IDF Train:', X_train.shape)
print('TF-IDF Test:', X_test.shape)
```

```python
# Train Naive Bayes model
nb = MultinomialNB(alpha=0.5)
nb.fit(X_train, y_train)

preds = nb.predict(X_test)
acc = accuracy_score(y_test, preds)
print('Naive Bayes Accuracy:', round(acc, 4))
print('\nClassification Report:')
print(classification_report(y_test, preds))
```

```python
# Save trained model
joblib.dump(nb, '../results/model_trained_files/naive_bayes_model.pkl')
print('Model saved to ../results/model_trained_files/naive_bayes_model.pkl')
```
