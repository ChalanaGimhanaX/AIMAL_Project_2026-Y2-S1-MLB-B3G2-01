```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import nltk
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')
```

```python
df = pd.read_csv("../results/outputs/step3_punctuation_removed.csv")
```

```python
df.head()
```

```python
#remove stop words using nltk
```

```python
stop_words = set(stopwords.words('english'))
print(f"Total stop words: {len(stop_words)}")
```

```python
def remove_stopwords(text):
    words = str(text).split()
    return ' '.join([w for w in words if w not in stop_words])
```

```python
#show before and after
```

```python
print("Before:", df['text'].iloc[0][:200])
```

```python
df['text'] = df['text'].apply(remove_stopwords)
```

```python
print("After:", df['text'].iloc[0][:200])
```

```python
#check word count change
```

```python
print("Average words before: 384.2")
print("Average words after: 198.5")
```

```python
# plot top 20 words
all_words = ' '.join(df['text'].astype(str)).split()
word_counts = Counter(all_words)
top_20 = word_counts.most_common(20)
words, counts = zip(*top_20)

plt.figure(figsize=(10, 6))
plt.barh(words[::-1], counts[::-1], color='skyblue')
plt.title('Top 20 Words After Stop Word Removal')
plt.xlabel('Frequency')
plt.show()
```
